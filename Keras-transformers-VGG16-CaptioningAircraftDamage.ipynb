```python
# checking if things are installed
import pkg_resources

required = {
    "pandas": "3.0.1",
    "tensorflow-cpu": "2.21.0",
    "pillow": "12.1.1",
    "matplotlib": "3.10.8",
    "transformers": "5.4.0",
    "torch": "2.11.0",
    "torchvision": "0.26.0",
    "torchaudio": "2.11.0"
}

for pkg, ver in required.items():
    try:
        installed = pkg_resources.get_distribution(pkg).version
        status = "OK" if installed == ver else f"WRONG VERSION ({installed})"
        print(f"{pkg}: {status}")
    except:
        print(f"{pkg}: NOT INSTALLED")
```
